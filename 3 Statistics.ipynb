A statistic is a quantity computed from a _data_ sample. 

Usually, a statistic tries to capture some simple information about the sample such as the location of centre of the distribution (e.g. mean, median) or its spread, scatter or dispersion (e.g. standard deviation). These are _summary statistics_.

Often statistics are chosen to be (unbiased) estimators of the properties of the population. For example the sample mean, $\bar{x}$, estimates the population mean $\mu$, or the sample variance, $s^2$, estimates the population variance $\sigma^2$.

## Common Univariate Statistics
For example, the following are common statistics based on a sample of $N$ measurements ${x_1 ... x_N}$.

The _sample mean_ is
$$
\bar{x} = \frac{\sum_i^N x_i}{N}
$$

The sample _median_ is obtained by rank ordering the data and choosing the middle value if there are an odd number, or the average of the middle two, if even. 

The _sample variance_, $s^2$, is
$$
s^2 \equiv {\rm Var}(x) = \frac{1}{N-1} \sum_i^N (x_i-\bar{x})^2 
$$

Notice the $N-1$ factor (sometimes called Bessel's correction) which applies when the mean is not known and is estimated from the data. 

For example, with a sample of one data point, and an unknown mean, we cannot estimate the variance.

Since statistics are functions of the random variables, there is a _standard error_ in each of these quantities. For example the standard error in the mean is:

$$
\sigma_{\bar{x}} = \frac{\sigma}{\sqrt{N}}
$$

If the population variance $\sigma^2$ is unknown, you can estimate the standard error by substituting the sample variance $s^2$.

The sample variance is also a statistic, so it has its own variance. 

If the sample is taken from a Normal distribution, then the variance of the sample variance is
$$
\sigma^2_{s^2} = s^4 \frac{2}{N-1}
$$

## Multivariate Statistics

The sample _covariance matrix_ is just the generalization of the sample variance.

Suppose I observe $k$ properties for each of $N$ objects.

This could be a list of catalogue properties, e.g. luminosity, mass, size, redshift etc. 

Or the properties could be closely related, for example, if each object has a spectrum, then the properties could be flux at wavelength 1, flux at wavelength 2 etc. 

I can write the data for a given object $i$ as a $k$-length column vector $\boldsymbol x_i$, and define the mean as:

$$
\boldsymbol{\overline{x}}= {1 \over {N}}\sum_{i=1}^N \boldsymbol x_i = \begin{bmatrix}
\bar{x}_1 \\
\vdots \\
\bar{x}_j \\ 
\vdots \\
\bar{x}_k
\end{bmatrix}
$$

In the spectrum example, $\boldsymbol{\overline{x}}$ would be the average spectrum.

and the covariance matrix as:
$$
\mathbf{C} = \frac{1}{N-1}\sum_{i=1}^N (\boldsymbol x_i- \boldsymbol{\overline{x}}) (\boldsymbol x_i-\boldsymbol{\overline{x}}^\mathrm{T})
$$

The diagonal elements are the variances of the individual components, and the offdiagonal elements describe the degree of correlation.

### Bivariate Statistics and Correlations

A common scenario is to focus only on two variables. Suppose we have $N$ data points each consisting of a pair 
($x_i$, $y_i$). Then the sample covariance matrix is just 

$$
\boldsymbol C = \frac{1}{N-1}
\begin{bmatrix}
\sum_i^N (x-\bar{x})^2 & \sum_i^N (x-\bar{x})(y-\bar{y})\\
\sum_i^N (x-\bar{x})(y-\bar{y}) & \sum_i^N (y-\bar{y})^2
\end{bmatrix}
$$

We can see that the diagonal elements are just the variances of $x$ and $y$ and must be positive or zero. The matrix is symmetric and the off-diagonal element can be positive or negative.

It is common to define the [Pearson correlation coefficient](https://en.wikipedia.org/wiki/Pearson_correlation_coefficient#For_a_sample) $r$ 
$$
r \equiv \frac{\sum^N _{i=1}(x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum^N _{i=1}(x_i - \bar{x})^2} \sqrt{\sum^N _{i=1}(y_i - \bar{y})^2}} = \frac{C_{12}}{\sqrt{C_{11}}\sqrt{C_{22}}}
$$

When $r = 0$ there is no correlation between $x$ and $y$: they are independent. In general $-1 \le r \le 1$


### Propagation of uncertainty

AKA Propagation of errors

#### Linear combinations of random variables

Suppose $z = ax + by +c$, and that $x$ and $y$ are random variables.

$$
\bar{z} = \overline{(ax + by +c)} = a \bar{x} + b \bar{y} + c 
$$


\begin{align*}
\mathrm{Var}(z) = \sigma^2_z &= \frac{1}{(N-1)} \sum_i^N (ax_i + by_i +c - a\bar{x} - b\bar{y} -c)^2 \\
&= \frac{1}{(N-1)} \sum_i^N \left[ a^2(x_i-\bar{x})^2 + b^2(y_i-\bar{y})^2 \right.\\
& ~~~~~~~~~~~~~~~~~~~~ \left. + 2 a b (x_i-\bar{x})(y_i-\bar{y}) \right]\\
&= a^2 \mathrm{Var}(x) + b^2 \mathrm{Var}(y) + 2ab\, \mathrm{Cov}(x,y) \\
&= a^2 \sigma^2_x + b^2 \sigma^2_y + 2ab\, \sigma_{xy}\\
\end{align*}

If $x$ and $y$ are independent, then $\mathrm{Cov}(x,y) = \sigma_{xy} = 0$ so the last term disappears and one gets the familiar rule learnt in e.g. physics labs.

Example: Difference between two independent measurements, $z = x - y$:

$$\sigma^2_z = \sigma_x^2 + \sigma_y^2$$

If, for example, $\sigma_y = 1/2 \sigma_x$, then $\sigma_z = 1.12 \sigma_x$. 

So this leads to a rule of thumb that if a source of error is $\lesssim 1/2$ of a more dominant source of error, you can probably neglect it.

Moral: don't waste a lot of effort "chasing down" small sources of error if you have much larger ones to deal with.
***

Example: if one is averaging N independent identical random variables, we find that $a = b = \ldots = 1/N$, so that

$$
\mathrm{Var}(\bar{x}) = \sigma^2_{\bar{x}} = N \times  (\frac{1}{N^2} \sigma^2) = \frac{\sigma^2}{N}
$$

as above

#### Nonlinear Combinations of Random Variables

Suppose we have some function $f(x,y)$, and that $x, y$ are a random variables.

If $\sigma_x \ll \bar{x}$ and likewise for $y$, we can Taylor expand f around the means 

$$
f \approx f(\bar{x},\bar{y}) +  \left(\frac{\partial f}{\partial {x}}\right)\left(x- \bar{x}\right) + 
\left(\frac{\partial f}{\partial{y}}\right)(y- \bar{y}) 
$$

where the partial derivatives are evaluated at their means, so these are constants.

The constants, such as $f(\bar{x},\bar{y})$, do not affect the variance so this behaves like the linear case above.

If $x$ and $y$ are independent, this gives
$$
\sigma^2_f = \left(\frac{\partial f}{\partial {x}}\right)^2 \sigma_x^2 + \left(\frac{\partial f}{\partial {y}}\right)^2 \sigma_y^2 
$$

and taking the square root gives the familiar formula.

***

__Example__: suppose $f = x^2$. Then the formula says $\sigma_f = 2 \bar{x} \sigma_x$, and so the fractional uncertainties are:

$$\left(\frac{\sigma_f}{\bar{f}}\right) = 2 \left(\frac{\sigma_x}{\bar{x}}\right)$$


#### Combining data with unequal measurement errors

When combining data with differing measurement errors $\sigma_i$, we are not obliged to do a straight average.

We can combine them with weights $w_i$, so long as the weights are normalized: $\sum_i^N w_i =1$. 

What is is the optimal choice of weights?

For a choice of weights, we can calculate the variance of the weighted mean using propagation of errors.

Minimizing this variance yields $w_i \propto 1/\sigma_i^2$ so that 

$$
\bar{x} = \frac{\sum_i^N \frac{x_i}{\sigma_i^2}}{\sum_i^N \frac{1}{\sigma_i^2}}
$$

Q: What is the standard error in the weighted mean? 

##### Example: Two measurements of the Hubble Constant:
    
Riess et al. 2021:  $73.2\pm1.3$
    
Planck: $67.66\pm0.42$

This is easy with the Python [uncertainties](https://pythonhosted.org/uncertainties/index.html) package. (You may need to install it with ``pip install uncertainties``)

```python
from uncertainties import ufloat

riess = ufloat(73.2, 1.3) #2021 update
print("Riess :", riess)
planck = ufloat(67.66, 0.42)
print("Planck:", planck)
```

```python
w1 = (riess.s)**(-2)
w2 = (planck.s)**(-2)
wmean = (w1*riess+w2*planck)/(w1+w2)

# you can control sig figs and formatting with the uncertainies package
print("The weighted mean is                               : {:.2u} with 2 sig figs in the error".format(wmean))
print("or or you can print all digits if you really want  :", wmean.n, "+/-", wmean.s)
print(f"or just the default rounding, but with LateX output: ${wmean:L}$")  # LaTeX
```

The uncertainty in the weighted mean is lower, as expected (although only very slightly, because the Planck value dominates)
