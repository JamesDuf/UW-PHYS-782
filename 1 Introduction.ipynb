## Why do we care about probability, statistics, data analysis and all that?

The scientific method is often summarized as a top-down process

1. Formulate a hypothesis that makes a _prediction_ for an observation or experiment
2. Gather data
3. Reject hypothesis, or fail to reject it.

It is at step 3 that we need to make a decision based on the data that we have. Understanding probability and statistics helps us make this decision. 

Aside: Of course, most of the time it doesn't work that way at all. 

Example:

> "I _predict_ that there are large amounts of dark matter in the Universe," 

said no theorist ever.

Dark matter was discovered serendipitously by studying the dynamics of clusters of galaxies (Zwicky in the 1930s) and of spiral galaxies (by Rubin and others in the 70s). 

It was at that time that theorists started making _post_-dictions (e.g. Cold Dark Matter). But these hypotheses led to new predictions.

So, in practice, science often (usually?) works very differently:

2. Gather data
3. Discover something interesting that one hadn't expected (at least according to 1.)
4. Decide whether the "interesting thing" merits further study*

*there are dangers with this approach that we will discuss later ...

For example, one might find that, unexpectedly, some parameter X is correlated with parameter Y. This in turn might suggest that X causes Y, or Y causes X or they are both caused by a third thing (Z). 

But first we need to decide whether the correlation is likely to be real or just a statistical fluke.

Either way, we need to make an informed _decision_ about what our data may be telling us. Understanding of probability and statistics will help us evaluate the odds for a given theory or model.

### Aside: Frequentists and Bayesians

There are two views about how to consider probabilities (and hence how one applies the scientific method)

From [this page](https://www.probabilisticworld.com/frequentist-bayesian-approaches-inferential-statistics/):

> According to the frequentist definition of probability, only repeatable random events (like the result of flipping a coin) have probabilities. These probabilities are equal to the long-term frequency of occurrence of the events in question. Frequentists don’t attach probabilities to hypotheses or to any fixed but unknown values in general.

> In contrast, Bayesians view probabilities as a more general concept. As a Bayesian, you can use probabilities to represent the uncertainty in any event or hypothesis. Here, it’s perfectly acceptable to assign probabilities to non-repeatable events, such as Kamala Harris winning the US presidential race in 2024.

> Orthodox frequentists would claim that such probabilities don’t make sense because the event is not repeatable. That is, you can’t run the election cycle an infinite number of times and calculate the proportion of them that Kamala Harris won.

We will generally be following the Bayesian approach. 

But there are cases where the frequentist approach makes sense.
