# Non-Parametric Statistics

## Parametric vs. Non-

The previous tests were _parametric_: we assumed the functional _form_ of the parent distribution (e.g. Gaussian).

We then measure some parameter(s) of the sample(s), such as means and variances, and compare those with the parent population.

What about cases where you do not know (or don't want to assume) the underlying functional form of the parent population?

This is where _non_-parametric statistics come in.

I will give two examples:

1. K-S test
2. Spearman rank correlation

## Kolmogorov-Smirnov (KS) test

Are two distributions equal (the null hypothesis)?

There are two versions: one where you compare a data sample to a theoretical distribution, and a second one where you compare two data samples.

### One sample KS test

1. From the data, calculate the observed cumulative probability, CDF$_{\rm obs}(x)$
<b>-> You do a histogram of the data and you order it in order, each bar contains its value and lower so can never go down (only up or flat)</b>
2. From theory do the same, yielding CDF$_{\rm theory}(x)$.
3. Calculate the maximum of the absolute value of differences as a function of $x$, $D$, between the two CDFs.

![KS_Example](https://upload.wikimedia.org/wikipedia/commons/c/cf/KS_Example.png) Credit: Wikipedia

How do you know if your $D$-value is too large? 

For small $N$, look up the critical values in a table. 

For large $N$, the critical value of $D$ for a $p$-value of 0.05 corresponds to $\sim 1.36/\sqrt{N}$.

### Two-sample K-S test

Suppose you have two sets of measurements and you want to know if they are drawn from the same parent distribution?

As before, but now calculate the CDF of each, and again, the maximum $D$ between them.

<b>Red curve would look like a step function as well now.</b>

Consult tables or ``scipy.stats.ks_2samp``

## Spearman Rank-Order Correlation

The Pearson $r$-statistic tests for a linear correlation. 

But what if the two variables are _correlated_ but not in a linear fashion?

A statistic that makes fewer assumptions about the underlying distribution is the _Spearman rank-order_ correlation coefficient.

Instead of using the actual values $(x_i,y_i)$ we will sort each and assign ranks $X_i$ and $Y_i$ (so that $1 < X_i < N$ and $1 < Y_i < N$). 

We then compare pairs of ranks ($X_i$, $Y_i$) instead of pairs of actual values.

Specifically, compute 

$$
r_s = 1 - 6\frac{\sum_{i=1}^N (X_i-Y_i)^2}{N^3-N}
$$

If the relation between $x$ and $y$ is monotonic, then $r_s$ will be $+1$ or $-1$.

![Spearman](https://upload.wikimedia.org/wikipedia/commons/thumb/4/4e/Spearman_fig1.svg/500px-Spearman_fig1.svg.png)
Credit: Wikipedia

In the limit of large $N$, $r_s$ can be tested the same way as the Pearson correlation coefficient, i.e. via the Student $t$-test.
