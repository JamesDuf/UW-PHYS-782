## Code up the Schechter Function

<mark>Write some code to calculate the [Schechter](https://web.mit.edu/physics/people/faculty/schechter_paul.html) Function and its integral.</mark> The Schechter Function gives the number of galaxies between $L$ and $L+dL$. 

$$
\phi(L) \  \mathrm{d}L = \phi_* \left(\frac{L}{L_*}\right)^\alpha \exp\left(-\frac{L}{L_*}\right) \frac{\mathrm{d}L}{L_*},
$$

The free parameters of the Schechter Function are: $L_*$, $\alpha$ and the normalization $\phi_*$.

Note that the Schechter Function is not probability distribution: the normalization can be dimensionless and represent, for example, the number of galaxies between $L$ and $L+dL$ in a cluster, or, more commonly have units of Mpc$^{-3}$ where it represents the space number density in the Universe.

<mark>You do not need to integrate this numerically</mark>: you can (and should) use the upper [Incomplete Gamma Function](https://en.wikipedia.org/wiki/Incomplete_gamma_function):

$$
\Gamma(a,x) = \int_x^{\infty} t^{a-1}\,e^{-t}\,{\rm d}t
$$

There are, however, two issues with Python implementation [scipy.special.gammaincc](https://docs.scipy.org/doc/scipy-0.14.0/reference/generated/scipy.special.gammaincc.html)

1. It is normalized differently than the Wikipedia definition:
``1 / gamma(a) * integral(exp(-t) * t**(a-1), t=x..inf)``

2. It does not handle cases where $a< 0$. However, the Wikipedia article provides a recurrence relation that you can code.

It's a little tedious, so you can use the code below.

```python
import scipy.special
#
def gammaincc(a,x):

    import math

    """
    Generalized complementary (upper) incomplete gamma function, using
    recurrence relations e.g. Wikipedia
    https://en.wikipedia.org/wiki/Incomplete_gamma_function
    
    In the notation of Wikipedia:
    Gamma(a,x) = [Gamma(a+1,x)-x**a*exp(-x)]/a

    But to be consistent with scipy special function, this version is normalized, i.e.
    it is Gamma(a,x)/Gamma(a)
    """

    if a < 0:
        # Note that because it is normalized and the recurrence
        # relations are for the unnormalized version, Gamma(a,x), we
        # need to add extra Gamma(a) factors
        return ((gammaincc(a+1.,x)*gamma(a+1.)-(x**a)*np.exp(-x))/a)/gamma(a)
    else:
        return scipy.special.gammaincc(a,x)

def gamma(a):
    if a < 0:
        return gamma(a+1.)/a
    else:
        return scipy.special.gamma(a)
```

## Simulate a Galaxy sample drawn from the Schechter Function

<mark>Generate a large sample $(\sim 100,000)$ of galaxies drawn from the Schechter function</mark> with $L_*=1\times 10^{10}L_{\odot}$ and $\alpha = -1.2$.  The normalization $\phi_*$ is not important: you can set it to 1.

The Schechter function diverges as $L \rightarrow 0$ so you will need to set a lower luminosity limit, $L_{\rm low}$. For this exercise, I suggest $0.001\times L_*$

Note that the Schechter function is a power law times an exponential. The power law part by itself is _integrable_, so you can easily <mark>draw random samples from a power law</mark>. Then you can <mark>use [rejection sampling](https://theclevermachine.wordpress.com/2012/09/10/rejection-sampling/) (with the power-law distribution as the envelope) to simulate the exponential cutoff</mark>.

<mark>Plot a histogram of the _log_ of counts in bins of _log_ L </mark>to check if your random Schechter function is working.

At low $L \ll L_*$ the power law should appear as a straight line in a log-log plot. At high $L$ you should see the exponential cutoff. See the examples in Lecture Notebook 13.

1) Code up Schecter function

```python
def Schechter(Lstar, alpha, phistar):
    x = L / Lstar

    value = phistar * x**alpha * np.exp(-x) * Lstar**(-1)

    return value
```

2) integral from L_low to infinity 

let $x = L / L_\star$ such that 
\begin{equation*}
    \int^{\infty}_{L_{\rm low}} \phi(L) \mathrm{d}L = \int^{\infty}_{L_{\rm low}} \phi_* \left(\frac{L}{L_*}\right)^\alpha \exp\left(-\frac{L}{L_*}\right) \frac{\mathrm{d}L}{L_*} = \int^{\infty}_{L_{\rm low}/L_\star}  \phi_\star x^\alpha e^{-x}\mathrm{d}x 
\end{equation*}
and since 
$$
\Gamma(a,b) = \int_b^{\infty} x^{a-1}\,e^{-x}\,{\rm d}x
$$
we pull out the constant $\phi_\star$ and say that
\begin{equation*}
  \phi_\star \int^{\infty}_{L_{\rm low}/L_\star}   x^\alpha e^{-x}{\rm d}x = \phi_\star \Gamma \big(\alpha+1, L_{\rm low}/L_\star \big)
\end{equation*}

```python
def SchechterIntegral(Llow, Lstar, alpha, phistar):
    a = alpha + 1
    x = Llow / Lstar

    value = phistar * ( gammaincc(a,x) * gamma(a) )       # Need to multiply by gamma(a) because the gammainc doc above says that it return Gamma(a,x) / Gamma(a) and we want only Gamma(a,x)

    return value 
```

3. Generate Samples

- Draw samples from the power law part $x^\alpha$ 
- Reject based on exp cutoff probability $e^{-x}$

```python
import astropy.units as u

Lstar = 1e10 * u.solLum
alpha = -1.2
phistar = 1 

Llow = 0.001 * Lstar

N = 100_000
```

I use the CDF of a power law with negative index: 
\begin{equation*}
    F(x) = 1 - \bigg( \frac{x}{x_{\rm low}} \bigg)^{\alpha+1 }
\end{equation*} 
such that if I set the CDF equal to a uniform random variable $u$ then I can rearrange and sample via: 
\begin{equation*}
    u = 1 - \bigg( \frac{x}{x_{\rm low}} \bigg)^{\alpha+1 }
\end{equation*}

\begin{equation*}
    \boxed{ x = x_{\rm low}(1-u)^{1/(1+\alpha)} }
\end{equation*}

```python
rng = np.random.default_rng(123)                # set seed

xlow = Llow / Lstar         # 0.001 
xhigh = 1e6                 # basically inf since x^1.2 goes to zero as x -> 0 

accepted = []

# loop until we have 100,000 valid gals 
while len(accepted) < N:                        
    u = rng.random(N)                                  # generate an array of N floats between 0.0 and 1.0 (uniform) 
    x_trial = xlow * (1 - u)**(1 / (alpha + 1))        # Draw trials from the power-law distribution
    
    accept_value = rng.random(N)                       # generate an array of N floats between 0.0 and 1.0 (uniform)
    keep = accept_value < np.exp(-x_trial)             # Accept each trial with probability exp(-x)

    accepted.extend(x_trial[keep])                     # Add only the accepted trials to our sample


x_sample = np.array(accepted[:N])               # Keep only N samples 
L_sample = x_sample * Lstar                     # convert x back to luminosity units 
print(f"Generated {len(L_sample)} galaxies")
```

```python
logL = np.log10(np.asarray(L_sample / Lstar))

counts, edges = np.histogram(logL, bins=60)
centres = (edges[:-1] + edges[1:]) / 2

# Zero counts cannot be shown on a logarithmic axis
nonzero = counts > 0

plt.plot(centres[nonzero], counts[nonzero])
plt.yscale("log")
plt.xlabel(r"$\log_{10}(L/L_*)$")
plt.ylabel("Counts")
plt.grid(alpha=0.3)
plt.show()
```

1. as L << Lstar, we do get straight line in log-log space -> power law
2. as L >> Lstar, we do get an exponential cutoff

## Real Data
Now apply this to a real data set: the SDSS Main Galaxy Sample.

For these data, note that $H_0 = 100$ km/s/Mpc has been assumed when calculating the absolute magnitudes. The absolute magnitudes also include $K$-corrections. All magnitudes have been corrected for Galactic extinction.

```python
import astropy.table

sdssa = astropy.table.Table.read('sdss.csv', format='ascii.csv')
print(sdssa.colnames)
```

* ``RA, DEC``: J2000 coordinates, in degrees
* ``m``: apparent magnitude in r band
* ``mmax``: largest (i.e. faintest) apparent magnitude to which that galaxy would have been included in the sample.
* ``ABSMu, ABSMr``: Absolute magnitude in $^{0.1}u$ and $^{0.1}r$ bands. These bands are shifted from the standard ones to match the filter transmission of a galaxy at redshift 0.1 through the standard bands
* ``Z``: redshift 

```python
# The file is huge, so let's work with a small slice on the sky
sdssra = np.logical_and(sdssa['RA'] > 135., sdssa['RA']<225.)
sdssdec = np.logical_and(sdssa['DEC'] > 0., sdssa['DEC']<6.)
sdss = sdssa[np.logical_and(sdssra, sdssdec)]
plt.plot(sdss['RA'],sdss['DEC'],'.')
plt.grid();
```

```python
from astropy.cosmology import FlatLambdaCDM
lcdm = FlatLambdaCDM(Om0=0.3, H0=100.)
d = lcdm.comoving_distance(sdss['Z']).value
```

Let's make a "pie diagram".

```python
fig = plt.figure(figsize=(18,9))
ax = fig.add_subplot(111)
x = d*np.cos((270-sdss['RA'])/180.*np.pi)
y = d*np.sin((270.-sdss['RA'])/180.*np.pi)
plt.plot(x,y,marker="o", lw=0, ms=8./fig.dpi);
ax.set_xlim(-600,600)
ax.set_ylim(0.,600)
plt.grid()
plt.xlabel('X')
plt.ylabel('Y')
plt.title("SDSS North Galactic 0 < Dec < 6");
```

Is that not beautiful? The structure at $X \sim 0$, $Y \sim 220$ is part of the Sloan Great Wall.

The usual conversion from apparent magnitudes to absolute magnitudes depends on the luminosity distance and the $K$-correction

$$
M = m - 5 \log_{10}\frac{{D_L}}{10 \mathrm{pc}} - K
$$

In this case the absolute magnitude and the apparent magnitude are given for each galaxy, so we know the value of 
$$
5 \log_{10}\frac{{D_L}}{10 \mathrm{pc}} + K
$$

This can be used to obtain the absolute magnitude corresponding to the apparent magnitude limits.


```python
import astropy.units as u
dl = lcdm.luminosity_distance(sdss['Z']) # calculate luminosity distance for LCDM
distmod = 5.*np.log10(dl/(10*u.pc)) # and the distance modulus

magdiff = sdss['m']-sdss['ABSMr']
plt.plot(sdss['Z'], magdiff-distmod,'.')
plt.xlabel('z')
plt.ylabel('K');
```
